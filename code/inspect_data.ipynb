```python
def plot_nii_intensity_histogram(fpath, bins=100, ignore_zeros=True, log_scale=False, mask_path=None, show_stats=True):
    """
    Plots intensity histogram for a .nii(.gz) image.
    Args:
        fpath (str): path to image file
        bins (int): number of histogram bins
        ignore_zeros (bool): exclude zero-intensity voxels (common background)
        log_scale (bool): use logarithmic y-axis
        mask_path (str or None): optional path to mask nii; if provided, histogram uses masked voxels
        show_stats (bool): overlay basic statistics on the plot
    """
    if not os.path.exists(fpath):
        print(f"File not found: {fpath}")
        return

    img = sitk.ReadImage(fpath)
    arr = sitk.GetArrayFromImage(img).astype(np.float32)  # z,y,x

    if mask_path is not None:
        if not os.path.exists(mask_path):
            print(f"Mask file not found: {mask_path}")
            return
        m = sitk.GetArrayFromImage(sitk.ReadImage(mask_path)).astype(bool)
        if m.shape != arr.shape:
            print(f"Mask shape {m.shape} != image shape {arr.shape}")
            return
        data = arr[m]
    else:
        data = arr.ravel()

    if ignore_zeros:
        data = data[data != 0]

    if data.size == 0:
        print("No voxels to plot after applying filters.")
        return

    # compute stats
    mn, mx = float(data.min()), float(data.max())
    mean, med, std = float(data.mean()), float(np.median(data)), float(data.std())
    voxels = data.size

    plt.figure(figsize=(8,4))
    plt.hist(data, bins=bins, color='C0', alpha=0.7)
    if log_scale:
        plt.yscale('log')
    plt.xlabel('Intensity')
    plt.ylabel('Voxel count')
    title = os.path.basename(fpath) + f" — voxels={voxels:,}"
    plt.title(title)

    if show_stats:
        stats_text = f"min={mn:.3g}\nmax={mx:.3g}\nmean={mean:.3g}\nmedian={med:.3g}\nstd={std:.3g}"
        plt.gca().text(0.98, 0.95, stats_text, transform=plt.gca().transAxes,
                       fontsize=9, va='top', ha='right',
                       bbox=dict(boxstyle='round', facecolor='white', alpha=0.8))

    plt.tight_layout()
    plt.show()

# Example usage:
plot_nii_intensity_histogram("/home/minhchau/anaconda3/envs/curvas/curvas/Data/CURVASPDAC_00003/image.nii.gz", bins=200, ignore_zeros=True, log_scale=False)
plot_nii_intensity_histogram("/home/minhchau/anaconda3/envs/curvas/curvas/Data/CURVASPDAC_00029/image.nii.gz", bins=200, ignore_zeros=True, log_scale=False)
plot_nii_intensity_histogram("/home/minhchau/anaconda3/envs/curvas/curvas/Data/CURVASPDAC_00044/image.nii.gz", bins=200, ignore_zeros=True, log_scale=False)
plot_nii_intensity_histogram("/home/minhchau/anaconda3/envs/curvas/curvas/Data/CURVASPDAC_00055/image.nii.gz", bins=200, ignore_zeros=True, log_scale=False)
plot_nii_intensity_histogram("/home/minhchau/anaconda3/envs/curvas/curvas/Data/CURVASPDAC_00079/image.nii.gz", bins=200, ignore_zeros=True, log_scale=False)
```

```python
def visualize_intensity_with_colorbar(fpath, slice_idx=None, cmap='viridis', 
                                      vmin=None, vmax=None, figsize=(12, 5)):
    """
    Hiển thị ảnh CT với colorbar thể hiện giá trị intensity tại từng pixel.
    
    Args:
        fpath: đường dẫn file .nii.gz
        slice_idx: index của slice (None = slice giữa)
        cmap: colormap ('viridis', 'jet', 'gray', 'hot', 'coolwarm')
        vmin, vmax: giới hạn intensity để hiển thị (None = auto)
        figsize: kích thước figure
    """
    if not os.path.exists(fpath):
        print(f"File not found: {fpath}")
        return
    
    img = sitk.ReadImage(fpath)
    arr = sitk.GetArrayFromImage(img)  # z, y, x
    
    if slice_idx is None:
        slice_idx = arr.shape[0] // 2
    
    slice_data = arr[slice_idx]
    
    fig, axes = plt.subplots(1, 2, figsize=figsize)
    
    # Ảnh gốc với grayscale
    axes[0].imshow(slice_data, cmap='gray')
    axes[0].set_title(f'Grayscale - Slice {slice_idx}')
    axes[0].axis('off')
    
    # Ảnh với colormap và colorbar
    im = axes[1].imshow(slice_data, cmap=cmap, vmin=vmin, vmax=vmax)
    axes[1].set_title(f'Intensity Map ({cmap})')
    axes[1].axis('off')
    cbar = plt.colorbar(im, ax=axes[1], fraction=0.046, pad=0.04)
    cbar.set_label('Hounsfield Units (HU)', rotation=270, labelpad=20)
    
    plt.tight_layout()
    plt.show()
    
    # In thống kê
    print(f"File: {os.path.basename(fpath)}")
    print(f"Slice {slice_idx}: min={slice_data.min():.1f} HU, max={slice_data.max():.1f} HU")
    print(f"Mean: {slice_data.mean():.1f} HU, Std: {slice_data.std():.1f} HU")


def interactive_intensity_explorer(fpath, slice_idx=None, window_size=5):
    """
    Hiển thị ảnh và cho phép click để xem giá trị intensity tại vị trí đó.
    
    Args:
        fpath: đường dẫn file .nii.gz
        slice_idx: index của slice (None = slice giữa)
        window_size: kích thước vùng xung quanh pixel để lấy trung bình
    """
    if not os.path.exists(fpath):
        print(f"File not found: {fpath}")
        return
    
    img = sitk.ReadImage(fpath)
    arr = sitk.GetArrayFromImage(img)  # z, y, x
    
    if slice_idx is None:
        slice_idx = arr.shape[0] // 2
    
    slice_data = arr[slice_idx]
    
    fig, ax = plt.subplots(figsize=(10, 8))
    im = ax.imshow(slice_data, cmap='gray', interpolation='nearest')
    ax.set_title(f'{os.path.basename(fpath)} - Slice {slice_idx}\nClick để xem intensity')
    plt.colorbar(im, ax=ax, label='HU')
    
    # Text annotation
    annot = ax.annotate("", xy=(0,0), xytext=(20,20), textcoords="offset points",
                        bbox=dict(boxstyle="round", fc="yellow", alpha=0.8),
                        arrowprops=dict(arrowstyle="->"))
    annot.set_visible(False)
    
    def on_click(event):
        if event.inaxes != ax:
            return
        
        x, y = int(event.xdata), int(event.ydata)
        
        if 0 <= y < slice_data.shape[0] and 0 <= x < slice_data.shape[1]:
            # Giá trị tại pixel
            value = slice_data[y, x]
            
            # Giá trị trung bình vùng xung quanh
            half_w = window_size // 2
            y_min, y_max = max(0, y-half_w), min(slice_data.shape[0], y+half_w+1)
            x_min, x_max = max(0, x-half_w), min(slice_data.shape[1], x+half_w+1)
            window_mean = slice_data[y_min:y_max, x_min:x_max].mean()
            
            # Cập nhật annotation
            annot.xy = (x, y)
            text = f"Pos: ({x}, {y})\nHU: {value:.1f}\nWindow avg: {window_mean:.1f}"
            annot.set_text(text)
            annot.set_visible(True)
            fig.canvas.draw_idle()
    
    fig.canvas.mpl_connect('button_press_event', on_click)
    plt.show()


# Ví dụ sử dụng:
print("=== 1. Visualize với colormap ===")
visualize_intensity_with_colorbar(
    "/home/minhchau/anaconda3/envs/curvas/curvas/Data/CURVASPDAC_00003/image.nii.gz",
    slice_idx=290,  # slice giữa
    cmap='jet',      # colormap: 'jet', 'viridis', 'hot', 'coolwarm'
    vmin=-200,       # giới hạn dưới
    vmax=300         # giới hạn trên
)

print("\n=== 2. Interactive explorer (click vào ảnh để xem HU) ===")
interactive_intensity_explorer(
    "/home/minhchau/anaconda3/envs/curvas/curvas/Data/CURVASPDAC_00003/image.nii.gz",
    slice_idx=None,
    window_size=5
)
```

## Chiến lược xử lý Z dimension không đồng nhất

### Vấn đề:
- Số slices từ 55-797 (chênh lệch ~14 lần!)
- Crop cứng về 128 → mất thông tin
- Pad lên 797 → lãng phí memory và computation

### Giải pháp tốt nhất cho pancreas segmentation:

#### **1. Crop around pancreas region (ROI-based)**
- Tụy chỉ chiếm ~30-50 slices giữa bụng
- Crop chỉ vùng chứa tụy + margin
- Giữ toàn bộ thông tin quan trọng

#### **2. Adaptive padding/cropping**
- Cases < 128 slices: pad
- Cases > 128 slices: crop vùng giữa (nơi có tụy)
- Target size: 128 slices (phù hợp với GPU 24GB)

#### **3. Multi-scale approach (advanced)**
- Downsample Z nhẹ (2×) thay vì crop
- Giữ được toàn bộ context
- Trade-off: mất resolution nhưng giữ được spatial info

```python
# Phân tích vị trí khối u trong Z dimension để quyết định crop strategy
import os
import SimpleITK as sitk
import numpy as np

data_root = "/home/minhchau/anaconda3/envs/curvas/curvas/Data"

print("Phân tích vị trí tumor trong Z dimension:\n")
tumor_stats = []

for case in sorted(os.listdir(data_root))[:5]:  # Test 5 cases đầu
    case_dir = os.path.join(data_root, case)
    if not os.path.isdir(case_dir):
        continue
    
    img_path = os.path.join(case_dir, "image.nii.gz")
    mask_path = os.path.join(case_dir, "annotation_staple.nii.gz")
    
    if not os.path.exists(img_path) or not os.path.exists(mask_path):
        continue
    
    # Đọc image và mask
    img = sitk.ReadImage(img_path)
    mask = sitk.ReadImage(mask_path)
    
    mask_arr = sitk.GetArrayFromImage(mask)  # (z, y, x)
    total_slices = mask_arr.shape[0]
    
    # Tìm slices có tumor
    slices_with_tumor = np.where(mask_arr.sum(axis=(1,2)) > 0)[0]
    
    if len(slices_with_tumor) > 0:
        tumor_start = slices_with_tumor[0]
        tumor_end = slices_with_tumor[-1]
        tumor_span = tumor_end - tumor_start + 1
        tumor_center = (tumor_start + tumor_end) // 2
        
        print(f"{case}:")
        print(f"  Total slices: {total_slices}")
        print(f"  Tumor slices: {tumor_start} → {tumor_end} (span={tumor_span})")
        print(f"  Tumor center: slice {tumor_center} ({tumor_center/total_slices*100:.1f}%)")
        print(f"  Tumor occupies: {tumor_span/total_slices*100:.1f}% of Z")
        
        tumor_stats.append({
            'case': case,
            'total': total_slices,
            'span': tumor_span,
            'center_ratio': tumor_center/total_slices
        })
    print()

# Kết luận
if tumor_stats:
    avg_span = np.mean([s['span'] for s in tumor_stats])
    avg_center = np.mean([s['center_ratio'] for s in tumor_stats])
    
    print("="*60)
    print("KẾT LUẬN:")
    print(f"Tumor trung bình chiếm: {avg_span:.1f} slices")
    print(f"Tumor thường ở vị trí: {avg_center*100:.1f}% của Z (gần giữa)")
    print("\nĐỀ XUẤT PREPROCESSING:")
    print(f"1. Crop vùng có tumor + margin 20 slices mỗi bên")
    print(f"2. Nếu > 128 slices: center crop 128")
    print(f"3. Nếu < 128 slices: pad về 128")
    print("="*60)
```

```python
def visualize_std_hu_map(fpath, slice_idx=None, window_size=5, figsize=(15, 5)):
    """
    Visualize Std HU map để thấy vùng không đồng nhất (heterogeneous) vs đồng nhất (homogeneous).
    
    Args:
        fpath: đường dẫn file .nii.gz
        slice_idx: index của slice (None = slice giữa)
        window_size: kích thước window để tính Std (5x5, 7x7, etc.)
        figsize: kích thước figure
    """
    if not os.path.exists(fpath):
        print(f"File not found: {fpath}")
        return
    
    img = sitk.ReadImage(fpath)
    arr = sitk.GetArrayFromImage(img)  # z, y, x
    
    if slice_idx is None:
        slice_idx = arr.shape[0] // 2
    
    slice_data = arr[slice_idx]
    
    # Tính Std map cho toàn bộ slice
    std_map = np.zeros_like(slice_data, dtype=np.float32)
    half_w = window_size // 2
    
    print(f"Computing Std HU map with {window_size}x{window_size} window...")
    for y in range(slice_data.shape[0]):
        for x in range(slice_data.shape[1]):
            y_min = max(0, y - half_w)
            y_max = min(slice_data.shape[0], y + half_w + 1)
            x_min = max(0, x - half_w)
            x_max = min(slice_data.shape[1], x + half_w + 1)
            
            window = slice_data[y_min:y_max, x_min:x_max]
            std_map[y, x] = np.std(window)
    
    # Visualization
    fig, axes = plt.subplots(1, 3, figsize=figsize)
    
    # 1. Original image
    axes[0].imshow(slice_data, cmap='gray')
    axes[0].set_title(f'Original CT - Slice {slice_idx}')
    axes[0].axis('off')
    
    # 2. Std HU map
    im1 = axes[1].imshow(std_map, cmap='jet', vmin=0, vmax=np.percentile(std_map, 95))
    axes[1].set_title(f'Std HU Map (window={window_size}x{window_size})')
    axes[1].axis('off')
    cbar1 = plt.colorbar(im1, ax=axes[1], fraction=0.046, pad=0.04)
    cbar1.set_label('Std HU', rotation=270, labelpad=20)
    
    # 3. Overlay
    axes[2].imshow(slice_data, cmap='gray', alpha=0.7)
    im2 = axes[2].imshow(std_map, cmap='jet', alpha=0.5, vmin=0, vmax=np.percentile(std_map, 95))
    axes[2].set_title('Overlay: Gray + Std HU')
    axes[2].axis('off')
    
    plt.tight_layout()
    plt.show()
    
    # Statistics
    print(f"\nStd HU Map Statistics:")
    print(f"  Min Std: {std_map.min():.2f} (homogeneous)")
    print(f"  Max Std: {std_map.max():.2f} (heterogeneous)")
    print(f"  Mean Std: {std_map.mean():.2f}")
    print(f"  Median Std: {np.median(std_map):.2f}")
    print(f"\nInterpretation:")
    print(f"  🟦 Blue/Green (Std < 5): Homogeneous tissue (tụy lành, gan)")
    print(f"  🟨 Yellow (Std 5-15): Moderately heterogeneous")
    print(f"  🟥 Red (Std > 15): Highly heterogeneous (tumor, vessels, boundaries)")


def compare_roi_std_hu(fpath, mask_path, slice_idx=None, window_size=5):
    """
    So sánh Std HU giữa vùng có tumor vs vùng không có tumor.
    
    Args:
        fpath: đường dẫn image .nii.gz
        mask_path: đường dẫn annotation .nii.gz
        slice_idx: index của slice (None = slice giữa có tumor)
        window_size: kích thước window để tính Std
    """
    if not os.path.exists(fpath) or not os.path.exists(mask_path):
        print("File not found!")
        return
    
    img = sitk.ReadImage(fpath)
    mask = sitk.ReadImage(mask_path)
    
    img_arr = sitk.GetArrayFromImage(img)
    mask_arr = sitk.GetArrayFromImage(mask)
    
    # Tìm slice có tumor nhiều nhất
    if slice_idx is None:
        tumor_counts = mask_arr.sum(axis=(1,2))
        slice_idx = np.argmax(tumor_counts)
        print(f"Auto-selected slice {slice_idx} (most tumor pixels)")
    
    slice_data = img_arr[slice_idx]
    slice_mask = mask_arr[slice_idx]
    
    # Tính Std map
    std_map = np.zeros_like(slice_data, dtype=np.float32)
    half_w = window_size // 2
    
    for y in range(slice_data.shape[0]):
        for x in range(slice_data.shape[1]):
            y_min = max(0, y - half_w)
            y_max = min(slice_data.shape[0], y + half_w + 1)
            x_min = max(0, x - half_w)
            x_max = min(slice_data.shape[1], x + half_w + 1)
            
            window = slice_data[y_min:y_max, x_min:x_max]
            std_map[y, x] = np.std(window)
    
    # Phân tích Std trong tumor vs non-tumor
    tumor_pixels = slice_mask > 0
    non_tumor_pixels = slice_mask == 0
    
    std_tumor = std_map[tumor_pixels]
    std_non_tumor = std_map[non_tumor_pixels]
    
    # Visualization
    fig, axes = plt.subplots(2, 2, figsize=(12, 10))
    
    # Original + mask
    axes[0, 0].imshow(slice_data, cmap='gray')
    axes[0, 0].contour(slice_mask, colors='red', linewidths=2)
    axes[0, 0].set_title('CT with Tumor Mask')
    axes[0, 0].axis('off')
    
    # Std map
    im = axes[0, 1].imshow(std_map, cmap='jet', vmin=0, vmax=np.percentile(std_map, 95))
    axes[0, 1].contour(slice_mask, colors='white', linewidths=2, linestyles='--')
    axes[0, 1].set_title('Std HU Map')
    axes[0, 1].axis('off')
    plt.colorbar(im, ax=axes[0, 1])
    
    # Histogram comparison
    axes[1, 0].hist(std_tumor, bins=50, alpha=0.7, label='Tumor', color='red', density=True)
    axes[1, 0].hist(std_non_tumor[::10], bins=50, alpha=0.7, label='Non-tumor', color='blue', density=True)
    axes[1, 0].set_xlabel('Std HU')
    axes[1, 0].set_ylabel('Density')
    axes[1, 0].set_title('Std HU Distribution')
    axes[1, 0].legend()
    axes[1, 0].grid(alpha=0.3)
    
    # Box plot
    axes[1, 1].boxplot([std_non_tumor[::10], std_tumor], labels=['Non-tumor', 'Tumor'])
    axes[1, 1].set_ylabel('Std HU')
    axes[1, 1].set_title('Std HU Comparison')
    axes[1, 1].grid(alpha=0.3)
    
    plt.tight_layout()
    plt.show()
    
    # Statistics
    print("\n" + "="*60)
    print("STD HU COMPARISON:")
    print(f"Tumor region:")
    print(f"  Mean Std: {std_tumor.mean():.2f}")
    print(f"  Median Std: {np.median(std_tumor):.2f}")
    print(f"  Std of Std: {std_tumor.std():.2f}")
    print(f"\nNon-tumor region:")
    print(f"  Mean Std: {std_non_tumor.mean():.2f}")
    print(f"  Median Std: {np.median(std_non_tumor):.2f}")
    print(f"  Std of Std: {std_non_tumor.std():.2f}")
    print(f"\nDifference:")
    print(f"  Tumor Std is {std_tumor.mean()/std_non_tumor.mean():.2f}x higher than non-tumor")
    print("="*60)


# Ví dụ sử dụng:
print("=== 1. Visualize Std HU Map ===")
visualize_std_hu_map(
    "/home/minhchau/anaconda3/envs/curvas/curvas/Data/CURVASPDAC_00029/image.nii.gz",
    slice_idx=44,
    window_size=7  # 7x7 window
)

print("\n=== 2. Compare Tumor vs Non-tumor Std HU ===")
compare_roi_std_hu(
    "/home/minhchau/anaconda3/envs/curvas/curvas/Data/CURVASPDAC_00029/image.nii.gz",
    "/home/minhchau/anaconda3/envs/curvas/curvas/Data/CURVASPDAC_00029/annotation_staple.nii.gz",
    window_size=7
)
```
